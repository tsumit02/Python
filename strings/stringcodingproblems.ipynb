# String Coding Problems

## 1. Count Vowels and Consonants

- Assume all characters are lowercase.
- The string may include special characters.

```python
s = '_hel lo 123$World'
c = 0
v =0

for ch in s:
    if ch.isalpha():
     if ch in 'aeiou':
        v +=1
     else:
        c +=1
     
print(f"There are this many {v} vowels and this many {c} consonents")
```

## 2. Count Character Frequencies

```python
s = 'banana'
s2 = 'malyalam'
```

### Using a Dictionary

```python
count = dict()
for ch in s:
    count[ch] = count.get(ch, 0)+1
count
```

## 3. Check for a Palindrome

```python
s = 'madam'

srev = s[::-1]

if s == srev:
    print("this is a Palindrome String")
else:
    print("Not a palindrome")
```

## 4. Check for Anagrams

- Return `True` if two strings contain the same characters in any order.
- Assume both strings have the same length and contain lowercase letters.

```python
s1 = 'listen'
s2 = 'silent'
```

### Time Complexity: $O(n \log n)$

```python
s1_sorted = sorted(s1)
s2_sorted = sorted(s2)

if s1_sorted == s2_sorted:
    print("Is Anagram")
else:
    print("Not an Anagram")
```
