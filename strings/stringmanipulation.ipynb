# String Manipulation

## 1. Case Conversion

- `lower()`: Converts the entire string to lowercase.
- `upper()`: Converts the entire string to uppercase.
- `title()`: Capitalizes the first letter of each word.
- `capitalize()`: Capitalizes only the first letter of the string.
- `swapcase()`: Swaps uppercase letters to lowercase and vice versa.

```python
str1 = 'this is DATA SCIENCE'
```

```python
str1.upper()
```

```python
str1.lower()
```

```python
str1.title()
```

```python
str1.capitalize()
```

```python
str1.swapcase()
```

## 2. Whitespace Handling

- `strip()`: Removes spaces or specified characters from both ends.
- `lstrip()` / `rstrip()`: Removes spaces or specified characters from the left or right side.

```python
str2 = '     this is DATA SCIENCE  '
```

```python
str2.strip()
```

```python
str2.lstrip()
```

```python
str2.rstrip()
```

```python
str2 = '#####this is a string$$$$####'
```

```python
str2.strip('#')
```

```python
str2.rstrip('#').rstrip('$')
```

## 3. Searching and Finding

- `find(sub)`: Returns the first index of a substring, or `-1` if it is not found.
- `index(sub)`: Works like `find()`, but raises `ValueError` if the substring is not found.
- `startswith()` / `endswith()`: Checks whether a string starts or ends with the given text.

```python
str1 = 'this is a data science'
```

```python
str1.find('data')
```

```python
str1.index('s')
```

```python
str1.startswith('th')
```

```python
str2.startswith('h')
```

## 4. Validation Checks

- `isdigit()`: Returns `True` if all characters are digits.
- `isalpha()`: Returns `True` if all characters are letters.
- `isalnum()`: Returns `True` if all characters are letters or digits.
- `islower()` / `isupper()`: Checks whether the string is lowercase or uppercase.
- `isspace()`: Returns `True` if the string contains only whitespace.

```python
str2 = "1234"
```

```python
str3 = 'abc123'
```

```python
str4 = 'vcdef'
```

```python
str2.isdigit()
```

```python
str3.isalnum()
```

```python
str4.isalpha()
```

```python
str1.isspace()
```

## 5. Splitting and Joining

- `split(sep)`: Splits a string into a list. The default separator is whitespace.
- `join(iterable)`: Joins iterable elements into one string using a separator.

```python
str1.split()
```

```python
str1.split('t')
```

```python
str2 = ['this', 'is', 'a', 'boy']
```

```python
' '.join(str2)
```
